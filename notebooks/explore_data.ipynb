# STAR Project – First Look at the Data

**Research question:** Does the benefit of small classes (K–3) last over time, up to grade 8 and high school?

**Data:** Project STAR, Tennessee 1985–1989. Source: Harvard Dataverse (hdl:1902.1/10766).

## Setup + Load data

```python
import pandas as pd
df = pd.read_spss("../data/raw/STAR_Students.sav")
print(df.shape)
df.head()
```

We have **11,601 students** and **379 variables**. Each row is one student.

## Variables per grade

Most column names start with the grade:
`gk` = kindergarten, `g1`–`g8` = grades 1–8, `hs` = high school.

Here we count how many columns each grade has.

```python
prefixes = ["gk", "g1", "g2", "g3", "g4", "g5", "g6", "g7", "g8", "hs"]

for p in prefixes:
    cols = [c for c in df.columns if c.lower().startswith(p)]
    print(p, len(cols))
```

## 4. Find the key variables

For our question we need two things:
1. **Class type** – was the student in a small or regular class?
2. **Math scores** – how well did the student do?

Here we search for column names that contain `classtype` or `math`.

```python
# class type in each grade (the treatment)
print([c for c in df.columns if "classtype" in c.lower()])
# math scores (the outcome)
print([c for c in df.columns if "math" in c.lower()])
```

### What we found

- **Class type** exists only for K–3 (`gkclasstype` … `g3classtype`). This is when the experiment ran.
- **Main math score:** `tmathss` (total math scaled score). It exists in every grade, K to 8.
- High school has no `tmathss`. It has SAT, ACT and GPA instead.
